```python
X_priority = df["Ticket Text"]
y_priority = df["Ticket Priority"]

X_train_priority, X_test_priority, y_train_priority, y_test_priority = train_test_split(
    X_priority,
    y_priority,
    test_size=0.2,
    random_state=42,
    stratify=y_priority
)

print("Training rows:", len(X_train_priority))
print("Testing rows:", len(X_test_priority))
print("\nTraining priority distribution:")
print(y_train_priority.value_counts())
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer

tfidf_priority = TfidfVectorizer(max_features=5000)

X_train_priority_tfidf = tfidf_priority.fit_transform(
    X_train_priority
)

X_test_priority_tfidf = tfidf_priority.transform(
    X_test_priority
)

print("Training shape:", X_train_priority_tfidf.shape)
print("Testing shape:", X_test_priority_tfidf.shape)
```

```python
from sklearn.linear_model import LogisticRegression

priority_model = LogisticRegression(max_iter=1000)

priority_model.fit(
    X_train_priority_tfidf,
    y_train_priority
)

y_pred_priority = priority_model.predict(
    X_test_priority_tfidf
)

print("Priority model trained successfully.")
```

```python
from sklearn.metrics import accuracy_score, classification_report

priority_accuracy = accuracy_score(
    y_test_priority,
    y_pred_priority
)

print("Priority accuracy:", round(priority_accuracy * 100, 2), "%")

print("\nClassification Report:")
print(classification_report(
    y_test_priority,
    y_pred_priority
))
```

```python
from sklearn.dummy import DummyClassifier
from sklearn.metrics import accuracy_score

priority_baseline = DummyClassifier(strategy="most_frequent")

priority_baseline.fit(
    X_train_priority_tfidf,
    y_train_priority
)

y_pred_priority_baseline = priority_baseline.predict(
    X_test_priority_tfidf
)

baseline_priority_accuracy = accuracy_score(
    y_test_priority,
    y_pred_priority_baseline
)

print(
    "Priority baseline accuracy:",
    round(baseline_priority_accuracy * 100, 2),
    "%"
)
```
